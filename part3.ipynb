```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.read_csv(r"C:\Users\Rahwa.B\Desktop\Analyst Lab\WA_FN.csv")

# Set a consistent figure size
plt.rcParams['figure.figsize'] = (8, 5)

department_attrition = pd.crosstab(
    df['Department'],
    df['Attrition'],
    normalize='index'
) * 100

department_attrition[['Yes', 'No']].plot(
    kind='bar',
    figsize=(8, 5)
)

plt.title('Employee Attrition by Department')
plt.xlabel('Department')
plt.ylabel('Percentage of Employees')
plt.xticks(rotation=0)
plt.legend(title='Attrition')
plt.tight_layout()
plt.show()
```

The chart compares employee attrition across departments. It shows the proportion of employees who stayed and those who left within each department. The department with the highest percentage of employees classified as Yes has the highest attrition rate. This comparison is more meaningful than simply counting the number of employees who left because departments may have different workforce sizes. Departments with relatively high attrition may require further investigation into factors such as workload, job satisfaction, management, compensation, and career opportunities.

```python
role_attrition = pd.crosstab(
    df['JobRole'],
    df['Attrition'],
    normalize='index'
) * 100

role_attrition['Yes'].sort_values().plot(
    kind='barh',
    figsize=(9, 6)
)

plt.title('Employee Attrition Rate by Job Role')
plt.xlabel('Attrition Rate (%)')
plt.ylabel('Job Role')
plt.tight_layout()
plt.show()
```

The chart shows that attrition varies considerably across job roles. Some roles like sales representative, laboratory technician, human resource have higher proportion of employees leaving than others. This suggests that job role may be an important factor associated with employee turnover. Roles with high attrition should be investigated further to determine whether factors such as compensation, workload, career progression, job satisfaction, or working conditions contribute to employee departures.

```python
overtime_attrition = pd.crosstab(
    df['OverTime'],
    df['Attrition'],
    normalize='index'
) * 100

overtime_attrition[['Yes', 'No']].plot(
    kind='bar',
    figsize=(8, 5)
)

plt.title('Employee Attrition by Overtime Status')
plt.xlabel('Overtime')
plt.ylabel('Percentage of Employees')
plt.xticks(rotation=0)
plt.legend(title='Attrition')
plt.tight_layout()
plt.show()
```

The visualization compares attrition between employees who work overtime and those who do not. The percentage of employees leaving is considerably higher among employees who work overtime, this suggests a potential relationship between overtime and employee attrition. Excessive working hours may contribute to workload pressure and reduced work-life balance. However, this chart shows an association rather than proving that overtime directly causes employees to leave.

```python
plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x='Age',
    hue='Attrition',
    bins=20,
    kde=True,
    multiple='layer'
)

plt.title('Age Distribution by Employee Attrition')
plt.xlabel('Age')
plt.ylabel('Number of Employees')
plt.tight_layout()
plt.show()
```

The histogram shows the distribution of employee ages for employees who stayed and employees who left. The two groups can be compared to identify differences in their age distributions. Employees who left are concentrated more heavily in younger age groups, this may indicate that younger employees experience higher turnover. This could potentially be related to career development, compensation, job mobility, or other factors. Further analysis would be required before concluding that age itself causes attrition.

```python
plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x='MonthlyIncome',
    hue='Attrition',
    bins=30,
    kde=True,
    multiple='layer'
)

plt.title('Monthly Income Distribution by Employee Attrition')
plt.xlabel('Monthly Income')
plt.ylabel('Number of Employees')
plt.tight_layout()
plt.show()
```

The histogram compares the distribution of monthly income between employees who stayed and employees who left. Employees who left are concentrated more heavily within lower income ranges, this may suggest that compensation is associated with employee attrition. Employees receiving higher salaries may show greater retention, although income can also be related to job level, experience, and job role. Therefore, income should be examined together with these other variables before drawing a final conclusion.

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x='Attrition',
    y='Age'
)

plt.title('Age Distribution by Employee Attrition')
plt.xlabel('Attrition')
plt.ylabel('Age')
plt.tight_layout()
plt.show()
```

The boxplot compares the age distribution of employees who stayed with those who left. The median age, interquartile range, and overall spread can be compared between the two groups. Employees who left have a lower median age than employees who stayed, this provides evidence of an association between younger age and higher attrition. The presence of overlapping distributions would indicate that age alone does not completely distinguish employees who leave from those who stay.

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x='Attrition',
    y='MonthlyIncome'
)

plt.title('Monthly Income Distribution by Employee Attrition')
plt.xlabel('Attrition')
plt.ylabel('Monthly Income')
plt.tight_layout()
plt.show()
```

The boxplot compares monthly income between employees who stayed and employees who left. It allows differences in median income, spread, and extreme values to be observed. The median income of employees who left is lower than that of employees who stayed, this suggests that lower income may be associated with higher attrition. However, income is strongly related to job level, role, and experience, so these variables should also be considered in further analysis.

```python
attrition_counts = df['Attrition'].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    attrition_counts,
    labels=attrition_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Overall Employee Attrition')
plt.show()
```

The pie chart shows the overall proportion of employees who stayed with the company compared with those who left. The No category represents employees who remained with the organization, while the Yes category represents employees who left.

```python
attrition_by_role = (
    df.groupby("JobRole")["Attrition"]
      .apply(lambda x: (x == "Yes").mean() * 100)
      .sort_values(ascending=False)
)

print(attrition_by_role)
```

```python
highest_role = attrition_by_role.idxmax()
highest_rate = attrition_by_role.max()

lowest_role = attrition_by_role.idxmin()
lowest_rate = attrition_by_role.min()

print(highest_role, highest_rate)
print(lowest_role, lowest_rate)
```

```python
attrition_by_department = (
    df.groupby("Department")["Attrition"]
      .apply(lambda x: (x == "Yes").mean() * 100)
      .sort_values(ascending=False)
)

print(attrition_by_department)
```

```python
highest_department = attrition_by_department.idxmax()
highest_rate = attrition_by_department.max()

lowest_department = attrition_by_department.idxmin()
lowest_rate = attrition_by_department.min()

print("Highest:", highest_department, highest_rate)
print("Lowest:", lowest_department, lowest_rate)
```
